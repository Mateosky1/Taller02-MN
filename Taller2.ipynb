# Verificar visualmente


```python
import numpy as np
from scipy.optimize import newton

def f(x):
    _y = x**3 - 3*x**2 + x - 1
    return _y

def fprime(x):
    _y = 3*x**2 - 6*x + 1
    return _y


raiz = newton(f, x0=100, fprime=fprime, tol=1e-6, maxiter=1000)

xs = list(np.linspace(-1, 5, 100))
ys = [f(x) for x in xs]

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.plot(xs, ys, label='f(x)', linewidth=2)
# Marcar la raíz con un punto rojo
plt.plot(raiz, 0, 'ro', markersize=10, label=f'Raíz ≈ {raiz:.4f}')
plt.axhline(y=0, color='k', linestyle='-', linewidth=0.5, alpha=0.3)
plt.axvline(x=raiz, color='r', linestyle='--', linewidth=1, alpha=0.5)
plt.xlabel('xs')
plt.ylabel('ys')
plt.title('xs vs ys - Método de Newton')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
import numpy as np
from scipy.optimize import newton
import matplotlib.pyplot as plt

puntos_x = []
puntos_y = []

def f(x):
    _y = x**3 - 3*x**2 + x - 1
    puntos_x.append(x)
    puntos_y.append(_y)
    return _y

def fprime(x):
    return 3*x**2 - 6*x + 1

raiz = newton(f, x0=100, fprime=fprime, tol=1e-6, maxiter=1000)

xs = np.linspace(-1, 105, 200)
ys = xs**3 - 3*xs**2 + xs - 1

plt.figure(figsize=(8, 5))
plt.plot(xs, ys, label='f(x)', linewidth=2, alpha=0.5)
plt.plot(puntos_x, puntos_y, 'go', markersize=6, label='Puntos evaluados')
plt.plot(raiz, 0, 'ro', markersize=10, label=f'Raiz ≈ {raiz:.4f}')
plt.axhline(y=0, color='k', linestyle='-', linewidth=0.5, alpha=0.3)
plt.axvline(x=raiz, color='r', linestyle='--', linewidth=1, alpha=0.5)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Metodo de Newton - Iteraciones')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

# Objetivo 6: Repetir el proceso para el Método de la Secante

```python
# Objetivo 6: Repetir el proceso para el Método de la Secante
puntos_x_sec = []
def f_secante(x):
    _y = x**3 - 3*x**2 + x - 1
    puntos_x_sec.append(x)
    print(f"Secante iteracion/evaluacion -> x = {x:.6f}, f(x) = {_y:.6f}")
    return _y

print("Proceso iterativo de la Secante:")
raiz_sec = newton(f_secante, x0=100, tol=1e-6, maxiter=1000)
print("-" * 60)
print(f"Raiz final encontrada (Secante): {raiz_sec:.6f}")

iteraciones_sec = range(1, len(puntos_x_sec) + 1)
plt.figure(figsize=(8, 5))
plt.plot(iteraciones_sec, puntos_x_sec, marker='s', linestyle='-', color='g')
plt.xlabel('Numero de Evaluacion')
plt.ylabel('Estimacion de la raiz (x)')
plt.title('Comportamiento de la Secante: Estimación de la raiz vs Iteracion')
plt.grid(True)
plt.show()
```

# Método de la Bisección

```python
from scipy.optimize import bisect

puntos_x_bis = []

def f_biseccion(x):
    _y = x**3 - 3*x**2 + x - 1
    puntos_x_bis.append(x)
    return _y

a, b = 2, 4
print(f"Calculando biseccion en el intervalo [{a}, {b}]...")

raiz_bis = bisect(f_biseccion, a, b, xtol=1e-6, maxiter=100)
print(f"Raiz encontrada con Biseccion: {raiz_bis:.6f}")
print(f"Evaluaciones necesarias: {len(puntos_x_bis)}")
```

* Graficar para la estimación x0=10, la tangente y las raíces durante la ejecución

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**3 - 3*x**2 + x - 1

def fprime(x):
    return 3*x**2 - 6*x + 1

x_tangentes = [10.0]
for _ in range(3):
    x_actual = x_tangentes[-1]
    x_siguiente = x_actual - f(x_actual)/fprime(x_actual)
    x_tangentes.append(x_siguiente)


xs_plot = np.linspace(1, 11, 500)
ys_plot = f(xs_plot)

plt.figure(figsize=(10, 6))
plt.plot(xs_plot, ys_plot, label='$f(x) = x^3 - 3x^2 + x - 1$', color='blue', linewidth=2)
plt.axhline(0, color='black', linewidth=1) # Corte con el eje X

colors = ['red', 'green', 'orange', 'purple']
for i in range(3):
    xk = x_tangentes[i]
    x_next = x_tangentes[i+1]
    
    # Ecuación de la recta tangente en geometría: y = f'(xk)*(x - xk) + f(xk)
    y_tan = fprime(xk) * (xs_plot - xk) + f(xk)
    plt.plot(xs_plot, y_tan, '--', color=colors[i], label=f'Tangente en $x_{i}={xk:.2f}$')
    
    # Puntos sobre la curva f(x) y las aproximaciones en el eje X
    plt.scatter([xk], [f(xk)], color=colors[i], zorder=5)
    plt.scatter([x_next], [0], color=colors[i], marker='x', s=100, zorder=5) # Raíz aproximada
    plt.plot([xk, xk], [0, f(xk)], ':', color='gray') # Línea de proyección vertical

plt.xlim(1, 11)
plt.ylim(-50, 750)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Método de Newton desde $x_0 = 10$: Líneas Tangentes y Aproximaciones')
plt.grid(True, alpha=0.5)
plt.legend()
plt.show()
```
