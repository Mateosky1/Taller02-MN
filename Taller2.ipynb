# Verificar visualmente


```python
import numpy as np
from scipy.optimize import newton

def f(x):
    _y = x**3 - 3*x**2 + x - 1
    return _y

def fprime(x):
    _y = 3*x**2 - 6*x + 1
    return _y


raiz = newton(f, x0=100, fprime=fprime, tol=1e-6, maxiter=1000)

xs = list(np.linspace(-1, 5, 100))
ys = [f(x) for x in xs]

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.plot(xs, ys, label='f(x)', linewidth=2)
# Marcar la raíz con un punto rojo
plt.plot(raiz, 0, 'ro', markersize=10, label=f'Raíz ≈ {raiz:.4f}')
plt.axhline(y=0, color='k', linestyle='-', linewidth=0.5, alpha=0.3)
plt.axvline(x=raiz, color='r', linestyle='--', linewidth=1, alpha=0.5)
plt.xlabel('xs')
plt.ylabel('ys')
plt.title('xs vs ys - Método de Newton')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
import numpy as np
from scipy.optimize import newton
import matplotlib.pyplot as plt

puntos_x = []
puntos_y = []

def f(x):
    _y = x**3 - 3*x**2 + x - 1
    puntos_x.append(x)
    puntos_y.append(_y)
    return _y

def fprime(x):
    return 3*x**2 - 6*x + 1

raiz = newton(f, x0=100, fprime=fprime, tol=1e-6, maxiter=1000)

xs = np.linspace(-1, 105, 200)
ys = xs**3 - 3*xs**2 + xs - 1

plt.figure(figsize=(8, 5))
plt.plot(xs, ys, label='f(x)', linewidth=2, alpha=0.5)
plt.plot(puntos_x, puntos_y, 'go', markersize=6, label='Puntos evaluados')
plt.plot(raiz, 0, 'ro', markersize=10, label=f'Raiz ≈ {raiz:.4f}')
plt.axhline(y=0, color='k', linestyle='-', linewidth=0.5, alpha=0.3)
plt.axvline(x=raiz, color='r', linestyle='--', linewidth=1, alpha=0.5)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Metodo de Newton - Iteraciones')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
# Objetivo 3: Graficar el comportamiento del algoritmo (raíz vs iteración) para Newton
iteraciones_newton = range(1, len(puntos_x) + 1)

plt.figure(figsize=(8, 5))
plt.plot(iteraciones_newton, puntos_x, marker='o', linestyle='-', color='b')
plt.xlabel('Numero de Iteracion')
plt.ylabel('Estimación de la raiz (x)')
plt.title('Comportamiento de Newton: Estimación de la raiz vs Iteracion')
plt.grid(True)
plt.show()
```

# Objetivo 4: Probar diferentes estimaciones iniciales

```python
# Objetivo 4: Probar diferentes estimaciones iniciales
estimaciones_iniciales = [100, 50, 10, 0, -10, -50]

print("Resultados con diferentes valores de x0 (Metodo de Newton):")
print("-" * 60)
for x0 in estimaciones_iniciales:
    try:
        raiz_est = newton(f, x0=x0, fprime=fprime, tol=1e-6, maxiter=100)
        print(f"Valor inicial x0 = {x0:4} -> Raiz convergente: {raiz_est:.6f}")
    except Exception as e:
        print(f"Valor inicial x0 = {x0:4} -> Falla o diverge.")
```

# Objetivo 6: Repetir el proceso para el Método de la Secante

```python
# Objetivo 6: Repetir el proceso para el Método de la Secante
puntos_x_sec = []
def f_secante(x):
    _y = x**3 - 3*x**2 + x - 1
    puntos_x_sec.append(x)
    print(f"Secante iteracion/evaluacion -> x = {x:.6f}, f(x) = {_y:.6f}")
    return _y

print("Proceso iterativo de la Secante:")
raiz_sec = newton(f_secante, x0=100, tol=1e-6, maxiter=1000)
print("-" * 60)
print(f"Raiz final encontrada (Secante): {raiz_sec:.6f}")

iteraciones_sec = range(1, len(puntos_x_sec) + 1)
plt.figure(figsize=(8, 5))
plt.plot(iteraciones_sec, puntos_x_sec, marker='s', linestyle='-', color='g')
plt.xlabel('Numero de Evaluacion')
plt.ylabel('Estimacion de la raiz (x)')
plt.title('Comportamiento de la Secante: Estimación de la raiz vs Iteracion')
plt.grid(True)
plt.show()
```

# Método de la Bisección

```python
from scipy.optimize import bisect

puntos_x_bis = []

def f_biseccion(x):
    _y = x**3 - 3*x**2 + x - 1
    puntos_x_bis.append(x)
    return _y

a, b = 2, 4
print(f"Calculando biseccion en el intervalo [{a}, {b}]...")

raiz_bis = bisect(f_biseccion, a, b, xtol=1e-6, maxiter=100)
print(f"Raiz encontrada con Biseccion: {raiz_bis:.6f}")
print(f"Evaluaciones necesarias: {len(puntos_x_bis)}")
```
